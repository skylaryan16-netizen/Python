Q. create an array of n dimensions ( taken from user) and create empty arrray. then create empty array and take input from user according the dimensions and print the filled array

np.ndindex(dims)
- Automatically genrate all index combinations
- works on any number of dimensions


```python
d1 = int(input("Enter dimensions (space separated): "))
d2 = int(input("Enter dimensions (space separated): "))

dim = (d1,d2)
```

```python
import numpy as np

# Step 1: Take dimensions from user

dims = tuple(map(int, input("Enter dimensions (space separated): ").split()))

# Step 2: Create empty array
arr = np.empty(dims)

# Step 3: Take input for each element

for index in np.ndindex(dims):
    arr[index] = float(input(f"Enter value at index {index}: "))

# Step 4: Print filled array
print("\nFilled Array:")
print(arr)   

```
